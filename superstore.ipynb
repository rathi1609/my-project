# Data Visualixation

```python
plt.figure(figsize = (8,5))
sns.countplot(x = 'Order Date', data = df)
plt.show()
```

```python
plt.figure(figsize = (5,4))
sns.countplot(x = 'Ship Date', data = df)
plt.show()
```

```python
order_count = df['Order Date'].count()
ship_count = df['Ship Date'].count()
plt.figure(figsize = (5,4))
plt.title('Date Count')
plt.bar(['Order Date', 'Ship Date'],[order_count, ship_count], edgecolor = 'k', width = 0.7)
plt.show()
```

```python
plt.figure(figsize=(15,8))
sns.barplot(data=df, x='Profit', y='Quantity')
plt.title('Profit vs Quantity')
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Sales'], bins=20, edgecolor='black')

plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(5,4))

plt.hist(df['Profit'], bins=20, edgecolor='black')

plt.title('PROFITS')
plt.xlabel('Profit')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,7))

plt.hist(df['Discount'], bins=20, edgecolor='black')

plt.title('Discount')
plt.xlabel('Discount')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x='Sales', y='Profit')

plt.title('Relationship between Sales and Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')

plt.show()
```

```python
category = df['Category'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    category,
    labels=category.index,
    autopct='%1.1f%%',
    startangle=90,
    explode=(0.05, 0.05, 0.05)
)

plt.title('Percentage of Orders by Category')

plt.show()
```

```python
segment = df['Segment'].value_counts()

plt.figure(figsize=(20,5))

plt.pie(
    segment,
    labels=segment.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage of Orders by Segment')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, y='Sales')

plt.title('Box Plot of Sales')
plt.ylabel('Sales')

plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Month'] = df['Order Date'].dt.to_period('M').astype(str)
monthly_sales = df.groupby('Month')['Sales'].sum().reset_index()


plt.figure(figsize=(12,6))

plt.plot(
    monthly_sales['Month'],
    monthly_sales['Sales'],
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```
